```python
from mmdet.apis import init_detector
from pathlib import Path
import glob

# 가중치 + config 위치 (zip 안 구조에 따라 조정)
work = Path('/content/work/model')
config_path = list(work.rglob('rtmdet_tiny_pill.py'))[0]
ckpt_path = list(work.rglob('best_coco_bbox_mAP_epoch_15.pth'))[0]

print(f"Config: {config_path}")
print(f"Ckpt:   {ckpt_path}")

# CPU 강제
model = init_detector(str(config_path), str(ckpt_path), device='cpu')
print("\n모델 로드 완료 (CPU)")
print(f"디바이스: {next(model.parameters()).device}")
```

```python
from pathlib import Path

# 평가셋 위치 확인
eval_root = Path('/content/work/eval')
imgs = sorted(list(eval_root.rglob('*.png')) + list(eval_root.rglob('*.jpg')))
print(f"평가셋 이미지: {len(imgs)}장")
print(f"샘플: {imgs[0].name}")
```

```python
from mmdet.apis import inference_detector
import time
import numpy as np

# Warm-up 5회 (첫 추론은 lazy init 때문에 느림)
print("Warm-up 5회...")
for _ in range(5):
    _ = inference_detector(model, str(imgs[0]))
print("Warm-up 완료\n")

# 본 측정: 33장 × 3 round = 99회
N_ROUNDS = 3
all_times = []

for round_idx in range(N_ROUNDS):
    round_times = []
    for img_path in imgs:
        t0 = time.perf_counter()
        _ = inference_detector(model, str(img_path))
        t1 = time.perf_counter()
        round_times.append((t1 - t0) * 1000)  # ms
    all_times.append(round_times)
    print(f"Round {round_idx+1}: 평균 {np.mean(round_times):.0f}ms / 장")

# 통계
flat = np.array(all_times).flatten()
print(f"\n=== 최종 결과 ===")
print(f"전체 측정: {len(flat)}회 (33장 × {N_ROUNDS} round)")
print(f"평균:    {np.mean(flat):.0f}ms = {np.mean(flat)/1000:.2f}초")
print(f"표준편차: {np.std(flat):.0f}ms")
print(f"min:    {np.min(flat):.0f}ms")
print(f"max:    {np.max(flat):.0f}ms")
print(f"median: {np.median(flat):.0f}ms")

# 판정 (v3.7 룰)
mean_sec = np.mean(flat) / 1000
print(f"\n=== 판정 (v3.7 6단계) ===")
if mean_sec <= 2:
    print(f"✅ 직접 배포 가능 (CPU 2초 이내)")
elif mean_sec <= 5:
    print(f"⚠️  데모 가능, 답답함, 최적화 권장 (CPU 2~5초)")
else:
    print(f"❌ 부적합, ONNX/양자화/2위 교체 필요 (CPU 5초 초과)")
```

```python
import json
from datetime import datetime

result = {
    'measured_at': datetime.now().strftime('%Y-%m-%d %H:%M'),
    'environment': 'Colab CPU runtime (2 vCPU, HF Spaces 무료 환경 근사)',
    'model': 'RTMDet-tiny (seed=100, best_coco_bbox_mAP_epoch_15.pth)',
    'imgsz': 640,
    'n_images': 33,
    'n_rounds': N_ROUNDS,
    'mean_ms': float(np.mean(flat)),
    'std_ms': float(np.std(flat)),
    'min_ms': float(np.min(flat)),
    'max_ms': float(np.max(flat)),
    'median_ms': float(np.median(flat)),
    'mean_sec': float(np.mean(flat)/1000),
    'verdict': '직접 배포 가능' if mean_sec <= 2 else ('최적화 권장' if mean_sec <= 5 else '부적합'),
}

# JSON 박제
out_json = '/content/work/cpu_inference_speed.json'
with open(out_json, 'w', encoding='utf-8') as f:
    json.dump(result, f, indent=2, ensure_ascii=False)

# Markdown 박제
md = f"""# CPU 추론 속도 측정 (v3.7 6단계)

측정: {result['measured_at']}
환경: {result['environment']}
모델: {result['model']}
imgsz: {result['imgsz']}

## 결과

| 지표 | 값 |
|---|---|
| 측정 횟수 | {result['n_images']}장 × {result['n_rounds']} round = {result['n_images']*result['n_rounds']}회 |
| 평균 | {result['mean_ms']:.0f}ms ({result['mean_sec']:.2f}초) |
| 표준편차 | {result['std_ms']:.0f}ms |
| min / median / max | {result['min_ms']:.0f} / {result['median_ms']:.0f} / {result['max_ms']:.0f}ms |

## 판정 (v3.7 룰)

- 2초 이내 → 직접 배포 가능
- 2~5초 → 데모 가능, 답답함, 최적화 권장
- 5초 초과 → 부적합

**판정: {result['verdict']}**
"""

out_md = '/content/work/cpu_inference_speed.md'
with open(out_md, 'w', encoding='utf-8') as f:
    f.write(md)

print(f"저장: {out_json}")
print(f"저장: {out_md}")
print("\n--- 내용 ---")
print(md)

# 다운로드
from google.colab import files
files.download(out_md)
files.download(out_json)
```
