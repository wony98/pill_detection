## 6. 데이터셋 다운로드 (gdown)

본인 Drive 공유 zip 두 개 다운로드 (약 4분, 75MB/s 측정값).
- `dataset_yolo.zip` (17.9GB) — K-code 학습 데이터
- `synthetic_combined.zip` (3.3GB) — 합성 통합본

gdown이 한도 차단(`FileURLRetrievalError: Too many users have viewed`) 뜨면 24시간 대기 또는 본인 Drive에 사본 만들어 새 file_id로 시도.

```python
import os, time, zipfile
from googleapiclient.http import MediaIoBaseDownload

DOWNLOADS = {
    'dataset_yolo.zip':  '1miPs4XRtLP1noLnfSQrbQyA-95XnKIM0',
    'synthetic_v3.zip':  '1mfX3SStw41M5qi24ss6VHo5ZvIKn11QG',
}

def verify_zip(path):
    """zip 무결성 검증 — testzip()이 None 반환하면 정상"""
    try:
        with zipfile.ZipFile(path) as zf:
            bad = zf.testzip()
            return bad is None, bad
    except Exception as e:
        return False, str(e)

for fname, fid in DOWNLOADS.items():
    out = f'/content/{fname}'

    # 이미 있으면 무결성 검증 후 스킵 판단
    if os.path.exists(out) and os.path.getsize(out) > 1e9:
        print(f'\n=== {fname}: 이미 있음 ({os.path.getsize(out)/1e9:.2f} GB), 무결성 검증 ===')
        ok, bad = verify_zip(out)
        if ok:
            print(f'  무결성 OK, 스킵')
            continue
        else:
            print(f'  손상 발견 ({bad}) → 삭제 후 재다운로드')
            os.remove(out)

    print(f'\n=== {fname} 다운로드 (Drive API, 1GB 청크) ===')
    request = drive_service.files().get_media(fileId=fid)

    t0 = time.time()
    last_t = t0
    last_bytes = 0

    with open(out, 'wb') as fp:
        downloader = MediaIoBaseDownload(fp, request, chunksize=1024*1024*1024)
        done = False
        chunk_idx = 0
        while not done:
            status, done = downloader.next_chunk()
            chunk_idx += 1
            now = time.time()
            current_bytes = fp.tell()
            chunk_mb = (current_bytes - last_bytes) / 1024 / 1024
            chunk_sec = now - last_t
            chunk_speed = chunk_mb / chunk_sec if chunk_sec > 0 else 0
            avg_speed = (current_bytes / 1024 / 1024) / (now - t0) if (now - t0) > 0 else 0
            pct = int(status.progress() * 100) if status else (100 if done else 0)
            print(f'  청크 {chunk_idx}: {pct}% | 이번 {chunk_mb:.0f}MB / {chunk_sec:.1f}s = {chunk_speed:.0f} MB/s | 평균 {avg_speed:.0f} MB/s')
            last_t = now
            last_bytes = current_bytes

    elapsed = time.time() - t0
    size_gb = os.path.getsize(out) / 1e9
    print(f'다운로드 완료: {size_gb:.2f} GB / {elapsed:.0f}초 / {size_gb*1024/elapsed:.0f} MB/s')

    # 다운로드 직후 무결성 검증
    print(f'무결성 검증 중...')
    ok, bad = verify_zip(out)
    if ok:
        print(f'  무결성 OK')
    else:
        raise RuntimeError(f'다운로드 직후에도 손상: {bad}. 재시도 필요.')

print('\n=== 모든 zip 다운로드 + 무결성 검증 완료 ===')
```

## 7. 압축 해제 + 카운트 검증

dataset_yolo (8302/2090) + synthetic_combined (7200/800) 둘 다 한 번에 풀고 검증.

```python
import os, zipfile, time, glob, shutil
from tqdm import tqdm

# zip별 압축 해제 사양
# - extract_to: zip을 풀 대상 디렉토리
# - skip_prefix: zip 내부에서 제외할 경로 prefix (None이면 전체)
# - expected_root: 압축 해제 후 생성될 최종 루트 (검증용)
EXTRACT_PLAN = [
    {
        'zip':           '/content/dataset_yolo.zip',
        'extract_to':    '/content',                # zip 내부 최상위가 dataset_yolo/
        'skip_prefix':   None,
        'expected_root': '/content/dataset_yolo',
    },
    {
        'zip':           '/content/synthetic_v3.zip',
        'extract_to':    '/content',                # zip 내부 최상위가 synthetic_v3/
        'skip_prefix':   'synthetic_v3/palm_test/', # palm 시각화용, 학습 불필요
        'expected_root': '/content/synthetic_v3',
    },
]

def has_train_val(root):
    """train/val 이미지가 모두 존재하면 True (압축 해제 완료 판단)"""
    img_train = os.path.join(root, 'images', 'train')
    img_val   = os.path.join(root, 'images', 'val')
    if not (os.path.isdir(img_train) and os.path.isdir(img_val)):
        return False
    return len(os.listdir(img_train)) > 0 and len(os.listdir(img_val)) > 0

for plan in EXTRACT_PLAN:
    zp           = plan['zip']
    extract_to   = plan['extract_to']
    skip_prefix  = plan['skip_prefix']
    expected     = plan['expected_root']
    name         = os.path.basename(expected)

    # 이미 정상적으로 풀렸는지 확인 (train/val 둘 다 비어있지 않아야 인정)
    if has_train_val(expected):
        print(f'{name}: 이미 풀려있음 (train/val 정상), 스킵')
        continue

    # 빈 폴더 또는 부분 압축 해제 흔적이 있으면 정리
    if os.path.exists(expected):
        print(f'{name}: 불완전한 잔재 발견, 삭제 후 재해제')
        shutil.rmtree(expected)

    print(f'\n=== {os.path.basename(zp)} 압축 해제 ===')
    t0 = time.time()
    with zipfile.ZipFile(zp) as zf:
        members = zf.namelist()
        if skip_prefix:
            members = [m for m in members if not m.startswith(skip_prefix)]
            print(f'  학습용 엔트리: {len(members):,} ({skip_prefix} 제외)')
        else:
            print(f'  엔트리: {len(members):,}')

        for m in tqdm(members, desc=f'  {name}'):
            zf.extract(m, extract_to)
    print(f'완료: {time.time()-t0:.0f}초')

# 카운트 검증 (v3 실제 카운트 기준)
KCODE = '/content/dataset_yolo'
SYNTH = '/content/synthetic_v3'
k_train = len(glob.glob(f'{KCODE}/images/train/*'))
k_val   = len(glob.glob(f'{KCODE}/images/val/*'))
s_train = len(glob.glob(f'{SYNTH}/images/train/*'))
s_val   = len(glob.glob(f'{SYNTH}/images/val/*'))

print(f'\n=== 카운트 검증 ===')
print(f'K-code train/val: {k_train} / {k_val}  (목표 8302/2090)')
print(f'Synth  train/val: {s_train} / {s_val}  (목표 6509/800)')

assert (k_train, k_val, s_train, s_val) == (8302, 2090, 6509, 800), \
    f'카운트 불일치: K-code({k_train},{k_val}) Synth({s_train},{s_val})'
print('\n카운트 OK — 학습 진행 가능')
```

## 9. YOLO → COCO 포맷 변환

K-code train + synthetic train 통합 → train_coco.json. 파일명 중복 회피를 위해 abs_path를 file_name으로 사용.

```python
import os, json, glob
from PIL import Image
from tqdm import tqdm

categories = [{'id': i+1, 'name': name, 'supercategory': 'pill'}
              for i, name in enumerate(CLASS_NAMES)]


def yolo_to_coco(image_dirs, label_dirs, output_path):
    images, annotations = [], []
    img_id, ann_id = 1, 1

    for img_dir, lbl_dir in zip(image_dirs, label_dirs):
        img_files = sorted(glob.glob(f'{img_dir}/*'))
        for img_path in tqdm(img_files, desc=os.path.basename(img_dir)):
            fname = os.path.basename(img_path)
            stem = os.path.splitext(fname)[0]
            lbl_path = f'{lbl_dir}/{stem}.txt'

            try:
                with Image.open(img_path) as im:
                    W, H = im.size
            except Exception as e:
                print(f'이미지 로드 실패: {img_path} - {e}')
                continue

            images.append({
                'id': img_id, 'file_name': fname,
                'width': W, 'height': H,
                'abs_path': img_path,
            })

            if os.path.exists(lbl_path):
                with open(lbl_path) as f:
                    for line in f:
                        parts = line.strip().split()
                        if len(parts) != 5:
                            continue
                        cls, cx, cy, w, h = map(float, parts)
                        bw, bh = w * W, h * H
                        bx = cx * W - bw / 2
                        by = cy * H - bh / 2
                        annotations.append({
                            'id': ann_id, 'image_id': img_id,
                            'category_id': int(cls) + 1,
                            'bbox': [bx, by, bw, bh],
                            'area': bw * bh, 'iscrowd': 0,
                        })
                        ann_id += 1
            img_id += 1

    coco = {'images': images, 'annotations': annotations, 'categories': categories}
    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(coco, f, ensure_ascii=False)
    print(f'\n저장: {output_path}')
    print(f'images: {len(images)}, annotations: {len(annotations)}')


# Train 통합
yolo_to_coco(
    image_dirs=['/content/dataset_yolo/images/train', '/content/synthetic_v3/images/train'],
    label_dirs=['/content/dataset_yolo/labels/train', '/content/synthetic_v3/labels/train'],
    output_path='/content/train_coco.json'
)

# Val 통합
yolo_to_coco(
    image_dirs=['/content/dataset_yolo/images/val', '/content/synthetic_v3/images/val'],
    label_dirs=['/content/dataset_yolo/labels/val', '/content/synthetic_v3/labels/val'],
    output_path='/content/val_coco.json'
)
```

## 10. file_name → abs_path 교체

mmdet이 절대경로로 이미지를 찾을 수 있도록 file_name을 abs_path로 교체.

```python
import json

for path in ['/content/train_coco.json', '/content/val_coco.json']:
    with open(path) as f:
        coco = json.load(f)

    # abs_path 처리는 이미 됐을 텐데 한 번 더 안전하게
    for img in coco['images']:
        if 'abs_path' in img:
            img['file_name'] = img.pop('abs_path')

    # info / licenses 필드 (pycocotools loadRes 호환)
    if 'info' not in coco:
        coco['info'] = {'description': 'Pill Detection 118 classes', 'version': '1.0', 'year': 2026}
    if 'licenses' not in coco:
        coco['licenses'] = []

    with open(path, 'w', encoding='utf-8') as f:
        json.dump(coco, f, ensure_ascii=False)
    print(f'{path} 패치 완료')
```

## 13. AutoUploadDriveHook 등록 + 학습 실행

매 epoch 끝나면 work_dir 통째로 본인 Drive 폴더에 자동 백업. 도중 끊겨도 셀 재실행 시 마지막 epoch에서 resume.

```python
import os, zipfile, time
from mmengine.config import Config
from mmengine.runner import Runner
from mmengine.hooks import Hook
from mmengine.registry import HOOKS


@HOOKS.register_module(force=True)
class AutoUploadDriveHook(Hook):
    """매 epoch 끝마다 work_dir 통째로 zip → Drive 본인 폴더에 직접 업로드"""
    priority = 'LOW'

    def __init__(self, work_dir, author, drive_folder_id, run_tag):
        self.work_dir = work_dir
        self.author = author
        self.drive_folder_id = drive_folder_id
        self.zip_name = f'{author}_{run_tag}_work_dir.zip'
        self.cached_file_id = None

    def _get_drive(self):
        from googleapiclient.discovery import build
        return build('drive', 'v3')

    def _find_existing(self, drive):
        results = drive.files().list(
            q=f"'{self.drive_folder_id}' in parents and name='{self.zip_name}' and trashed=false",
            fields='files(id)'
        ).execute()
        files = results.get('files', [])
        return files[0]['id'] if files else None

    def after_train_epoch(self, runner):
        from googleapiclient.http import MediaFileUpload

        zip_path = f'/content/{self.zip_name}'
        if os.path.exists(zip_path):
            os.remove(zip_path)

        runner.logger.info(f'[Upload] zip 생성 시작')
        t0 = time.time()
        with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_STORED) as zf:
            for root, _, fs in os.walk(self.work_dir):
                for f in fs:
                    fp = os.path.join(root, f)
                    arc = os.path.relpath(fp, self.work_dir)
                    zf.write(fp, arc)
        size_mb = os.path.getsize(zip_path) / 1e6
        runner.logger.info(f'[Upload] zip 생성 완료 ({size_mb:.0f} MB, {time.time()-t0:.0f}초)')

        try:
            drive = self._get_drive()
            if self.cached_file_id is None:
                self.cached_file_id = self._find_existing(drive)

            media = MediaFileUpload(zip_path, resumable=True, chunksize=50*1024*1024)

            if self.cached_file_id:
                request = drive.files().update(
                    fileId=self.cached_file_id, media_body=media, fields='id,name')
            else:
                metadata = {'name': self.zip_name, 'parents': [self.drive_folder_id]}
                request = drive.files().create(
                    body=metadata, media_body=media, fields='id,name')

            t0 = time.time()
            response = None
            while response is None:
                status, response = request.next_chunk()
            runner.logger.info(f'[Upload] Drive 업로드 완료 ({time.time()-t0:.0f}초)')

            if not self.cached_file_id:
                self.cached_file_id = response['id']

        except Exception as e:
            runner.logger.warning(f'[Upload] 실패: {e}')


# config 로드 + custom hook 등록
cfg = Config.fromfile(f'{WORK_DIR}/rtmdet_tiny_pill.py')
cfg.custom_hooks = list(cfg.get('custom_hooks', [])) + [
    dict(type='AutoUploadDriveHook',
         work_dir=WORK_DIR,
         author=AUTHOR,
         drive_folder_id=MY_FOLDER_ID,
         run_tag=RUN_TAG)
]
# resume 처리
last_ckpt_file = os.path.join(WORK_DIR, 'last_checkpoint')
if os.path.exists(last_ckpt_file):
    with open(last_ckpt_file) as f:
        ckpt_path = f.read().strip()
    print(f'resume 모드: {os.path.basename(ckpt_path)}')
    cfg.resume = True
else:
    print(f'첫 학습 시작 (load_from: {os.path.basename(cfg.load_from)})')

runner = Runner.from_cfg(cfg)
runner.train()

print('\n=== 학습 종료 ===')
```

## 17. Top-3 그룹핑 함수

같은 박스에 여러 클래스 후보가 있을 때 conf 기준으로 Top-3 추출. IoU 그룹핑(greedy) + 클래스 중복 제거.

```python
import numpy as np


def iou_xyxy(b1, b2):
    """xyxy 형식 IoU."""
    x1, y1 = max(b1[0], b2[0]), max(b1[1], b2[1])
    x2, y2 = min(b1[2], b2[2]), min(b1[3], b2[3])
    inter = max(0, x2 - x1) * max(0, y2 - y1)
    a1 = (b1[2] - b1[0]) * (b1[3] - b1[1])
    a2 = (b2[2] - b2[0]) * (b2[3] - b2[1])
    return inter / (a1 + a2 - inter + 1e-9)


def extract_top3(bboxes, scores, labels, class_names,
                 conf_thr=0.05, iou_thr=0.5, topk=3):
    keep = scores >= conf_thr
    bboxes, scores, labels = bboxes[keep], scores[keep], labels[keep]
    if len(bboxes) == 0:
        return []

    order = np.argsort(-scores)
    bboxes, scores, labels = bboxes[order], scores[order], labels[order]

    n = len(bboxes)
    group_id = -np.ones(n, dtype=int)
    next_gid = 0
    for i in range(n):
        if group_id[i] >= 0:
            continue
        group_id[i] = next_gid
        for j in range(i + 1, n):
            if group_id[j] >= 0:
                continue
            if iou_xyxy(bboxes[i], bboxes[j]) > iou_thr:
                group_id[j] = next_gid
        next_gid += 1

    results = []
    for gid in range(next_gid):
        mask = group_id == gid
        g_bboxes, g_scores, g_labels = bboxes[mask], scores[mask], labels[mask]

        seen_cls = {}
        for b, s, l in zip(g_bboxes, g_scores, g_labels):
            l = int(l)
            if l not in seen_cls or s > seen_cls[l][1]:
                seen_cls[l] = (b, float(s), l)

        cls_sorted = sorted(seen_cls.values(), key=lambda x: -x[1])[:topk]
        rep_bbox = cls_sorted[0][0].tolist()

        results.append({
            'bbox': rep_bbox,
            'top3_class_name': [class_names[l] for _, _, l in cls_sorted],
            'top3_confidence': [s for _, s, _ in cls_sorted],
        })
    return results


# sanity
_b = np.array([[10,10,50,50], [12,12,48,48], [200,200,250,250]], dtype=float)
_s = np.array([0.9, 0.7, 0.6])
_l = np.array([0, 1, 2])
_names = [f'class_{i}' for i in range(118)]
_out = extract_top3(_b, _s, _l, _names)
print(f'입력 박스 3 → 그룹 {len(_out)}개 (앞 두 박스 IoU 그룹핑되어 1그룹, 세 번째 별도 1그룹)')
```

## 20. 채점 + 결과 출력

GT bbox별 IoU≥0.5 매칭 후 Top-1 / Top-3 / Detection rate 계산.

```python
def evaluate_v4_json(output_json, iou_thr=0.5):
    n_top1 = n_top3 = n_match = 0
    img_top1_full = img_top3_full = 0
    n_false_pos = 0
    total_gt = sum(len(r['gt_bboxes']) for r in output_json['results'])
    n_imgs = len(output_json['results'])

    for r in output_json['results']:
        gts = r['gt_bboxes']
        preds = r['predictions']
        used = set()
        img_all_top1 = bool(gts)
        img_all_top3 = bool(gts)

        for gt in gts:
            best_iou, best_idx = 0.0, -1
            for j, p in enumerate(preds):
                if j in used:
                    continue
                i = iou_xyxy(gt['bbox'], p['bbox'])
                if i > best_iou:
                    best_iou, best_idx = i, j

            if best_iou >= iou_thr and best_idx >= 0:
                n_match += 1
                used.add(best_idx)
                top3 = [nfc(n) for n in preds[best_idx]['top3_class_name']]
                gt_name = nfc(gt['dl_name'])
                if top3 and top3[0] == gt_name:
                    n_top1 += 1
                else:
                    img_all_top1 = False
                if gt_name in top3:
                    n_top3 += 1
                else:
                    img_all_top3 = False
            else:
                img_all_top1 = False
                img_all_top3 = False

        if img_all_top1:
            img_top1_full += 1
        if img_all_top3:
            img_top3_full += 1
        n_false_pos += len(preds) - len(used)

    return {
        'detection_rate': n_match / total_gt,
        'top1_accuracy':  n_top1  / total_gt,
        'top3_accuracy':  n_top3  / total_gt,
        'image_top1':     img_top1_full / n_imgs,
        'image_top3':     img_top3_full / n_imgs,
        'n_top1': n_top1, 'n_top3': n_top3, 'n_match': n_match,
        'n_false_pos': n_false_pos, 'total_gt': total_gt, 'n_imgs': n_imgs,
    }


metrics = evaluate_v4_json(output_json)

print('=' * 50)
print(f'  RTMDet-tiny synth_e{best_epoch} seed=100 채점 결과')
print('=' * 50)
print(f"Detection (IoU>=0.5): {metrics['n_match']}/{metrics['total_gt']} ({metrics['detection_rate']:.1%})")
print(f"Top-1 정답:           {metrics['n_top1']}/{metrics['total_gt']} ({metrics['top1_accuracy']:.1%})")
print(f"Top-3 정답:           {metrics['n_top3']}/{metrics['total_gt']} ({metrics['top3_accuracy']:.1%})")
print(f"이미지 단위 Top-1:    {int(metrics['image_top1']*metrics['n_imgs'])}/{metrics['n_imgs']} ({metrics['image_top1']:.1%})")
print(f"이미지 단위 Top-3:    {int(metrics['image_top3']*metrics['n_imgs'])}/{metrics['n_imgs']} ({metrics['image_top3']:.1%})")
print(f"오탐 그룹:            {metrics['n_false_pos']}")
print(f"평균 추론 시간:       {output_json['ms_per_image']:.1f} ms/img")
```

## 21. v3 학습 곡선 시각화

scalars.json 파싱하여 train loss + val mAP epoch별 추이 시각화. Phase A-1 분석 패턴 그대로.

```python
import json
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# 한글 폰트 (없어도 figure 생성은 됨, 제목 깨질 뿐)
try:
    plt.rcParams['font.family'] = 'NanumGothic'
except:
    pass
plt.rcParams['axes.unicode_minus'] = False

# scalars.json 자동 탐색 (resume 시 여러 timestamp 폴더 생성 가능 → 최신 사용)
scalars_paths = sorted(glob.glob(f'{WORK_DIR}/**/vis_data/scalars.json', recursive=True))
assert len(scalars_paths) > 0, f"scalars.json 없음: {WORK_DIR}"
scalars_path = scalars_paths[-1]
print(f"scalars.json: {scalars_path}")

# 라인별 JSON 파싱
records = []
with open(scalars_path) as f:
    for line in f:
        line = line.strip()
        if line:
            records.append(json.loads(line))

df = pd.DataFrame(records)
print(f"전체 레코드: {len(df)}")

# train/val 분리 (키 존재 여부로)
train_df = df[df['loss'].notna()].copy() if 'loss' in df.columns else pd.DataFrame()
val_df = df[df['coco/bbox_mAP'].notna()].copy() if 'coco/bbox_mAP' in df.columns else pd.DataFrame()
print(f"train 레코드: {len(train_df)}, val 레코드: {len(val_df)}")

# train epoch별 평균 (iter 단위 → epoch 평균)
train_epoch_mean = train_df.groupby('epoch').agg({
    'loss': 'mean',
    'loss_cls': 'mean',
    'loss_bbox': 'mean'
}).reset_index()

# val: step 컬럼 = epoch 번호 (val_interval=1 가정)
# mmengine 내부 epoch 컬럼은 0/1만 있어 사용 불가 — 협업 원칙 89
val_df['epoch_real'] = val_df['step']

# best epoch (mAP@50:95 기준 = save_best='auto')
best_idx = val_df['coco/bbox_mAP'].idxmax()
best_epoch = int(val_df.loc[best_idx, 'epoch_real'])
best_map = val_df.loc[best_idx, 'coco/bbox_mAP']
final_map = val_df['coco/bbox_mAP'].iloc[-1]

print(f"\n=== v3 학습 곡선 요약 ===")
print(f"Best epoch:     {best_epoch} (mAP@50:95 = {best_map:.4f})")
print(f"Final epoch:    {int(val_df['epoch_real'].iloc[-1])} (mAP@50:95 = {final_map:.4f})")
print(f"Train loss e1:  {train_epoch_mean['loss'].iloc[0]:.4f}")
print(f"Train loss e10: {train_epoch_mean[train_epoch_mean['epoch']==10]['loss'].values[0]:.4f}")
print(f"Train loss e13: {train_epoch_mean['loss'].iloc[-1]:.4f}")

# 시각화: 1x2 subplot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 좌: train loss (총합 + cls + bbox)
ax = axes[0]
ax.plot(train_epoch_mean['epoch'], train_epoch_mean['loss'],
        'o-', label='loss (total)', linewidth=2)
ax.plot(train_epoch_mean['epoch'], train_epoch_mean['loss_cls'],
        's-', label='loss_cls', linewidth=1.5, alpha=0.8)
ax.plot(train_epoch_mean['epoch'], train_epoch_mean['loss_bbox'],
        '^-', label='loss_bbox', linewidth=1.5, alpha=0.8)
ax.axvline(10, color='gray', linestyle='--', alpha=0.5,
           label='switch_epoch=10 (stage2 start)')
ax.set_xlabel('Epoch')
ax.set_ylabel('Loss (epoch mean)')
ax.set_title('v3 Train Loss')
ax.legend()
ax.grid(alpha=0.3)

# 우: val mAP@50 + mAP@50:95
ax = axes[1]
ax.plot(val_df['epoch_real'], val_df['coco/bbox_mAP_50'],
        'o-', label='mAP@50', linewidth=2)
ax.plot(val_df['epoch_real'], val_df['coco/bbox_mAP'],
        's-', label='mAP@50:95', linewidth=2)
ax.axvline(best_epoch, color='red', linestyle='--', alpha=0.7,
           label=f'best epoch ({best_epoch})')
ax.axvline(10, color='gray', linestyle='--', alpha=0.5)
ax.set_xlabel('Epoch')
ax.set_ylabel('Val mAP')
ax.set_title('v3 Val mAP')
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
out_path = '/content/v3_training_curves.png'
plt.savefig(out_path, dpi=120, bbox_inches='tight')
plt.show()
print(f"\nfigure 저장: {out_path}")

# scalars.json도 같은 폴더로 복사 (로컬 다운로드용)
import shutil
shutil.copy(scalars_path, '/content/v3_scalars.json')
print(f"scalars.json 복사: /content/v3_scalars.json")
```
